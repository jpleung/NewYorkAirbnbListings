```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('new_york_listings_2024.csv', encoding_errors = 'ignore')
```

**Data Cleaning**

```python
data.isnull().sum()

# Dropping all rows with missing values
data.dropna(inplace=True)

data.isnull().sum()
```

```python
data.shape
```

```python
# Dealing with duplicate rows
data.duplicated().sum()

# Deleting all duplicated rows
data.drop_duplicates(inplace=True)

data[data.duplicated()]
```

```python
# Type casting
# Changing data types
data.dtypes

data['id'] = data['id'].astype(object)
data.dtypes

data['host_id'] = data['host_id'].astype(object)
data.dtypes
```

#### EDA
**Data Analysis**

**Univariate Analysis**

```python
# Identifying outliers in price

df = data[data['price'] < 1500]

sns.boxplot(data=df, x='price')
```

```python
# Price distribution

plt.figure(figsize=(8,5))
sns.histplot(data=df, x='price',bins=100)
plt.title('Price Distribution')
plt.ylabel('frequency')
plt.show()
```

```python
df.dtypes
```

```python
# Price distribution

plt.figure(figsize=(6,3))
sns.histplot(data=df, x='availability_365')
plt.title('Avialiability Distribution')
plt.ylabel('frequency')
plt.show()
```

```python
data.dtypes
```

```python
df.groupby(by='neighbourhood_group')['price'].mean()
```

```python
# Average price per bed

df.groupby(by='neighbourhood_group')['price_per_bed'].mean()
```

**Feature Engineering**

```python
# Create price_per_bed

df['price_per_bed'] = df['price']/df['beds']
df.head()
```

**Bivariate Analysis**

```python
df.columns
```

```python
# Price dependency by neighbourhood

sns.barplot(data=df, x='neighbourhood_group', y='price', hue='room_type')
```

```python
# Number of reviews and price relationship

plt.figure(figsize=(8,5))
plt.title('Locality and Review Dependency')
sns.scatterplot(data=df, x='number_of_reviews', y='price', hue='neighbourhood_group')
plt.show()
```

```python
df.dtypes
```

```python
sns.pairplot(data=df, vars=['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue='room_type')
```

```python
# Geographical distribution of Airbnb listings

plt.figure(figsize=(10,7))
sns.scatterplot(data=df, x='longitude', y='latitude', hue='room_type')
plt.title('Geographical Distribution of Airbnb Listings')
plt.show()
```

```python
df.dtypes
```

```python
# Heat map - correlation of one variable with others for numerical column

corr = df[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'availability_365', 'beds']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(data=corr, annot=True)
```
